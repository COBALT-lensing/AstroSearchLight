```python
from pathlib import Path
from matplotlib import pyplot

from resnet import ResNetFilter
from lightcurves import plot, load
```

```python
input_path = Path('data/tess_s56.txt')
with open(input_path, "r") as input_file:
    paths = input_file.readlines()
    paths = [p.strip() for p in paths[:100]]
```

```python
rn = ResNetFilter()
classifications = rn.classify_paths(paths)
```

```python
rn.classes
```

```python
for path, classification in classifications.items():
    print(*([path] + classification))
    lc = load(path)
    f = plot(lc)
    pyplot.show()
    pyplot.close(f)
```
