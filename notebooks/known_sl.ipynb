```python
from matplotlib import pyplot

from searchlight.lightcurves import get_spoc_urls, plot, load_tess
from searchlight.models.simulens import SimuLens
```

```python
# https://arxiv.org/abs/2410.13939
FALSE_POSITIVES = [273695675, 55276301, 289952607, 162565254, 13983380, 365950480, 455471849]
CANDIDATES = [
    82324681,
    320193438,
    247719397,
    61845524,
    457200146,
    314487191,
    155813746,
    468473101,
    229510522,
    394019762,
    65191369,
    247480759,
    230114166,
    124508784,
    299891691,
    9293371,
    369000430,
    29850601,
    307364849,
    234632282,
    151937435,
    453368802,
    293164329,
    238004509,
    320193939,
    63119966,
]

# https://arxiv.org/abs/2401.01477
KNOWN = ["KIC 12254688"]
```

```python
rn = SimuLens()
```

```python
dict([(1,45), (5, 66)])
```

```python
def classify_tics(tics):
    lc_urls = {}
    for tic in tics:
        lc_urls.update({u: tic for u in get_spoc_urls(tic)})
    return rn.classify_paths(list(lc_urls.keys())), lc_urls

def plot_results(results, tic_map):
    totals = {'lensing': 0, 'not-lensing': 0}
    for url, res in results.items():
        print(tic_map[url], res)
        lc = load_tess(url)
        plot(lc)
        pyplot.show()
        totals[res[0]] += 1
    return totals
```

# Known SL Systems

```python
results, tic_map = classify_tics(KNOWN)
```

```python
totals = plot_results(results, tic_map)
```
